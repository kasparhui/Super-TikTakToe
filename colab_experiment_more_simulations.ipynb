# Super TikTakToe: the test "more_simulations" on Google Colab

This continues the finished **large** network (iteration 240) with one thing changed: **800 simulations per move instead of 400**.
The finished network is not touched; the test works in its own folder
`Super-TikTakToe-models/experiments/more_simulations` in your Google Drive.

1. The training of large must have ended at iteration 240 ("Finished: ..."), and nothing may have been trained after that.
2. Menu **Runtime > Change runtime type**: choose the L4.
3. Run the cells from top to bottom. The first start copies the network and its memory (a few minutes) and plays
   the last 4 checks of large again with 512 games, as the baseline.
4. It runs 32 iterations of about 9 minutes, so about 5 hours (about 9 to 10 compute units), and ends by itself
   with a comparison "before the test" against "its last 3 checks".
5. After a disconnect run all cells again: the test continues where it stopped.
6. When it has ended: **Runtime > Disconnect and delete runtime**.

The other test has its own notebook and can run at the same time in a second session.
